## Goal
Use survey data to learn about students
interests and reasons for taking computing classes.

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the cleaned survey data
survey = pd.read_csv("non_majors_2024_cleaned.csv")

# Show the first five responses
survey.head()
```

## Where do students most often get information about computing courses at CCM?

```python
# Choose the source columns
sources = [
    "source_high_school",
    "source_it_website",
    "source_admissions",
    "source_advisor",
    "source_employer",
    "source_other"
]

# Count Yes answers for each source
source_counts = (survey[sources] == "Yes").sum()

# Sort from most to least 
source_counts = source_counts.sort_values(ascending=False)

source_counts
```

```python
# Make a bar chart
source_counts.plot(kind="barh")

plt.title("Sources of Computing Course Information")
plt.xlabel("Number of Students Who Answered Yes")
plt.ylabel("Information Source")
plt.show()
```

CCM advisors were the most common source of information, with 33 students answering Yes.

## Which computing topics are students most interested in?

```python
# Select the computing topic columns
topics = [
    "web_development",
    "cybersecurity",
    "app_development",
    "data_analytics",
    "machine_learning_ai",
    "programming",
    "game_design",
    "hardware"
]

# Count Yes answers and sort from most to least
topic_counts = (survey[topics] == "Yes").sum()
topic_counts = topic_counts.sort_values(ascending=False)

topic_counts
```

```python
# Compare interest in each topic with bar chart
topic_counts.plot(kind="barh")

plt.title("Computing Topics Students Are Interested In")
plt.xlabel("Number of Students Who Answered Yes")
plt.ylabel("Computing Topic")
plt.show()
```

Cybersecurity had the most interest, with 31 students answering Yes. Machine learning/AI and game design tied for second with 28 each


## What are the most common reasons students take computing classes?

```python
# Select the reasons for taking computing classes
reasons = [
    "required_class",
    "keep_skills_current",
    "career_advancement",
    "career_change",
    "professional_development",
    "job_displacement",
    "relocation",
    "it_certifications",
    "financial",
    "personal_enrichment",
    "curiosity"
]

# Count Yes answers and sort from most to least
reason_counts = (survey[reasons] == "Yes").sum()
reason_counts = reason_counts.sort_values(ascending=False)

reason_counts
```

```python
# Compare the reasons in a bar chart
reason_counts.plot(kind="barh")

plt.title("Reasons Students Take Computing Classes")
plt.xlabel("Number of Students Who Answered Yes")
plt.ylabel("Reason")
plt.show()
```

The most common reason was that the class was required for a degree, with 82 students answering Yes. Keeping skills current came next with 70, followed by professional development with 65.
